#### ANALYSIS BY SEX & TICKET CLASS

```python
import seaborn as sns
titanic = sns.load_dataset('titanic')

# Tỉ lệ sống sót
survival_combined = titanic.groupby(['sex', 'pclass'])['survived'].mean()

print("--- Tỉ lệ sống sót ---")
print(survival_combined.unstack() * 100)
```

Asssess:
- Tỉ lệ sống sót của phụ nữ có vé hạn 1 cao nhất gần 97%.
- Tỉ lệ sống sót của nam giới ở tất cả các hạng vé đều rất thấp, cho thấy rằng giới tính là một yếu tố quyết định đến khả năng sống sót của hành khách trên tàu Titanic.

#### C1 - Combined survival rate by Age Group and Ticket Class

```python
titanic_temp = titanic.copy()

# Chia nhóm tuổi: Trẻ em (0-12), Trưởng thành (13-59), Người già (60-100)
age_bins = [0, 12, 59, 100]
age_labels = ['Trẻ em', 'Trưởng thành', 'Người già']
titanic_temp['age_group'] = pd.cut(titanic_temp['age'], bins=age_bins, labels=age_labels)

# Groupby theo Hạng vé và Nhóm tuổi
survival_temp = titanic_temp.groupby(['pclass', 'age_group'], observed=False)['survived'].mean().unstack()

print("--- TỈ LỆ SỐNG SÓT THEO HẠNG VÉ VÀ NHÓM TUỔI ---")
print(survival_temp*100)
```

Assess:
- Xu hướng tăng rõ rệt: Đúng như dự đoán, tỉ lệ sống sót của mọi nhóm tuổi đều tăng mạnh khi chuyển từ hạng vé thấp (Hạng 3) lên hạng vé cao (Hạng 1).
- Note: Đặc quyền của trẻ em - Trẻ em ở khoang Hạng 1 và Hạng 2 được bảo vệ gần như tuyệt đối (tỉ lệ sống sót 100% ở hạng 2 và ~75.7% ở hạng 1). Tuy nhiên, trẻ em ở khoang Hạng 3 lại chỉ có khoảng ~41% sống sót. 
- Hạng vé có thể là yếu tố quyết định cơ hội sống của mọi lứa tuổi.

```python
# Lọc danh sách khách hàng có giá vé bằng 0
zero_fare = titanic[titanic['fare'] == 0]

# Tính toán các chỉ số
total_passengers = len(zero_fare)
gender_counts = zero_fare['sex'].value_counts().to_string(header=False)
min_age = zero_fare['age'].min()
max_age = zero_fare['age'].max()
survival_rate = zero_fare['survived'].mean()

# Hiển thị kết quả
print("--- THỐNG KÊ KHÁCH HÀNG VÉ 0 ĐỒNG ---")
print(f"- Tổng số lượng : {total_passengers} người")
print(f"- Giới tính     : {gender_counts.replace(chr(10), chr(10))}")
print(f"- Độ tuổi       : {min_age:.1f} - {max_age:.1f} tuổi")
print(f"- Tỉ lệ sống sót: {survival_rate:.2%}")
```

Assess:
- Họ là: Dữ liệu cho thấy có tổng cộng 15 người có vé 0$ trên tập dữ liệu này. Và 100% trong số họ là nam giới trong độ tuổi lao động (từ 19 đến 49 tuổi). Có thể là các thủy thủ đoàn, nhân viên bảo hành tàu hoặc những người được cấp vé danh dự
- Họ có được ưu tiên không? KHÔNG. Tỉ lệ sống sót của nhóm này cực kỳ thê thảm, chỉ đạt 6.67% (chỉ có 1/15 người sống sót).
