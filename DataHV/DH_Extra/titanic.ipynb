### 1. Load dataset Titanic bằng seaborn và in ra 5 dòng đầu. Tổng cộng có bao nhiêu hành khách? (Sử dụng Data/DH_w1/train.csv)


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Đọc dữ liệu từ file csv theo yêu cầu
df = pd.read_csv('../../Data/DH_w1/train.csv')
display(df.head())
print(f"Tổng cộng có {len(df)} hành khách.")
```

### 2. Dùng df.info() để xem kiểu dữ liệu từng cột. Cột nào có giá trị null? Cột nào nhiều null nhất?


```python
df.info()
null_counts = df.isnull().sum()
print("\nCác cột có giá trị null:")
print(null_counts[null_counts > 0])
print(f"\nCột nhiều null nhất là: {null_counts.idxmax()} với {null_counts.max()} giá trị null.")
```

### 3. Dùng df.describe() xem thống kê mô tả. Tuổi trung bình của hành khách là bao nhiêu? Giá vé cao nhất là bao nhiêu?


```python
display(df.describe())
print(f"Tuổi trung bình của hành khách là: {df['Age'].mean():.2f}")
print(f"Giá vé cao nhất là: {df['Fare'].max():.2f}")
```

### 4. Có bao nhiêu hành khách nam và bao nhiêu hành khách nữ? Dùng value_counts().


```python
gender_counts = df['Sex'].value_counts()
print(gender_counts)
```

### 5. Hành khách đi hạng mấy (1, 2, 3) nhiều nhất? Tính số lượng và tỉ lệ phần trăm mỗi hạng.


```python
class_counts = df['Pclass'].value_counts()
class_percentages = df['Pclass'].value_counts(normalize=True) * 100

print("Số lượng hành khách mỗi hạng:")
print(class_counts)
print("\nTỉ lệ phần trăm mỗi hạng:")
print(class_percentages)
print(f"\nHành khách đi hạng {class_counts.idxmax()} nhiều nhất.")
```

### 6. Lọc ra tất cả hành khách nữ. Họ chiếm bao nhiêu % tổng số hành khách?


```python
female_df = df[df['Sex'] == 'female']
female_percentage = (len(female_df) / len(df)) * 100
print(f"Hành khách nữ chiếm {female_percentage:.2f}% tổng số hành khách.")
```

### 7. Lọc ra hành khách đi hạng 1 (first class). Giá vé trung bình của nhóm này là bao nhiêu?


```python
first_class_df = df[df['Pclass'] == 1]
print(f"Giá vé trung bình của nhóm hạng 1 là: {first_class_df['Fare'].mean():.2f}")
```

### 8. Có bao nhiêu hành khách dưới 18 tuổi? Bao nhiêu người trên 60 tuổi?


```python
under_18 = df[df['Age'] < 18]
over_60 = df[df['Age'] > 60]
print(f"Số hành khách dưới 18 tuổi: {len(under_18)}")
print(f"Số hành khách trên 60 tuổi: {len(over_60)}")
```

### 9. Tính tỉ lệ sống sót của hành khách đi hạng 3 (third class). So sánh với tỉ lệ sống sót của hành khách đi hạng 1.


```python
survival_rate_3 = df[df['Pclass'] == 3]['Survived'].mean() * 100
survival_rate_1 = df[df['Pclass'] == 1]['Survived'].mean() * 100

print(f"Tỉ lệ sống sót hạng 3: {survival_rate_3:.2f}%")
print(f"Tỉ lệ sống sót hạng 1: {survival_rate_1:.2f}%")
print("Tỉ lệ sống sót của hành khách hạng 1 cao hơn nhiều so với hạng 3.")
```

### 10. Tính tỉ lệ sống sót của hành khách nữ và nam. So sánh xem nhóm nào có tỉ lệ sống sót cao hơn.


```python
survival_rate_female = df[df['Sex'] == 'female']['Survived'].mean() * 100
survival_rate_male = df[df['Sex'] == 'male']['Survived'].mean() * 100

print(f"Tỉ lệ sống sót của nữ: {survival_rate_female:.2f}%")
print(f"Tỉ lệ sống sót của nam: {survival_rate_male:.2f}%")
print("Hành khách nữ có tỉ lệ sống sót cao hơn nhiều so với nam.")
```

### 11. Vẽ biểu đồ cột (bar plot) thể hiện số lượng hành khách sống sót và không sống sót.


```python
sns.countplot(data=df, x='Survived')
plt.title('Số lượng hành khách sống sót và không sống sót')
plt.xlabel('Sống sót (0 = Không, 1 = Có)')
plt.ylabel('Số lượng')
plt.show()
```

### 12. Vẽ biểu đồ histogram thể hiện phân bố tuổi của hành khách.


```python
sns.histplot(data=df, x='Age', bins=30, kde=True)
plt.title('Phân bố tuổi của hành khách')
plt.xlabel('Tuổi')
plt.ylabel('Tần suất')
plt.show()
```
