```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Check the dataset
df = pd.read_csv('../../Data/Extra/salary_survey_raw.csv')
df.head()
```

Câu 1: Các em thấy dataset này có bao nhiêu vấn đề? Liệt kê không giới hạn số lượng.

```python
# DUPLICATE
print("---DUPLICATE---")
duplicate_rows = df.duplicated().sum()
print(f"Số lượng dòng trùng lặp: {duplicate_rows}")

# OUTLIER
print("\n---OUTLIER---")
plt.boxplot(df["Age"].dropna()) # dropna để boxplot không báo lỗi nếu Age có null
plt.show()

# MISSING VALUE
print("\n---MISSING VALUE---")
print("Giá trị thiếu: \n", df.isnull().sum())
print("\nTổng số giá trị thiếu:", df.isnull().sum().sum())

# Hướng dẫn tìm: Null CẦN XEM XÉT (Dựa vào tỷ lệ %)
print("\n---ĐÁNH GIÁ MỨC ĐỘ NULL (XEM XÉT ĐỂ DROP HAY FILL)---")
missing_ratio = (df.isnull().sum() / len(df)) * 100
print("Tỷ lệ thiếu (%) trên từng cột:")
print(missing_ratio[missing_ratio > 0].sort_values(ascending=False))

investigate_cols = missing_ratio[missing_ratio > 50].index.tolist()
if investigate_cols:
    print(f"\n=> ⚠️ CẢNH BÁO: Các cột sau bị Null quá 50% (Thuộc nhóm CẦN XEM XÉT XÓA HAY HỎI LẠI NGHIỆP VỤ): {investigate_cols}")

# INCONSISTENT DATA
print("\n---INCONSISTENT DATA---")
for col in df.select_dtypes(include=['object']).columns:
    print(f"[{col}] Unique Labels: {df[col].dropna().unique()[:10]}") # In thử 10 giá trị xem có lỗi chính tả không
#
df.describe()
```
