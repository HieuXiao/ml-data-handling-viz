### 1. Phân biệt Batch Learning vs Online Learning — chọn đúng cho bài toán

**Batch Learning (Học theo lô):**
- **Đặc điểm:** Hệ thống không thể học từ dữ liệu đang chảy vào theo thời gian thực (incremental learning). Nó phải được huấn luyện bằng toàn bộ dữ liệu có sẵn. Thường mất nhiều thời gian và tài nguyên tính toán.
- **Khi nào dùng:** Khi không cần hệ thống phải cập nhật tức thì, dữ liệu ít khi thay đổi nhanh chóng, và có đủ thời gian cũng như tài nguyên để huấn luyện định kỳ (VD: nhận diện hình ảnh, dự đoán doanh thu hàng tháng).

**Online Learning (Học trực tuyến):**
- **Đặc điểm:** Hệ thống học dần dần từ các luồng dữ liệu mới được đưa vào tuần tự (từng mẫu một hoặc các lô nhỏ gọi là mini-batches). Mô hình có thể cập nhật nhanh chóng ngay khi có dữ liệu mới.
- **Khi nào dùng:** Khi dữ liệu thay đổi liên tục và hệ thống cần thích nghi nhanh với xu hướng mới (VD: dự đoán giá cổ phiếu, phát hiện gian lận thẻ tín dụng, hệ thống gợi ý realtime), hoặc khi dữ liệu quá lớn không thể fit vào bộ nhớ (out-of-core learning).

### 2. Phân biệt Instance-Based vs Model-Based — hiểu trade-off memory vs speed

**Instance-Based Learning (Học dựa trên mẫu):**
- **Cách hoạt động:** Hệ thống học bằng cách ghi nhớ toàn bộ các ví dụ huấn luyện. Khi có dữ liệu mới, nó so sánh (dựa trên độ tương đồng/khoảng cách) với các mẫu đã nhớ để đưa ra dự đoán (VD: K-Nearest Neighbors).
- **Trade-off:** 
  - **Memory (Bộ nhớ):** Tốn cực kì nhiều bộ nhớ vì phải lưu trữ toàn bộ dữ liệu huấn luyện.
  - **Speed (Tốc độ):** Huấn luyện cực kì nhanh (vì chỉ cần lưu lại dữ liệu) nhưng dự đoán rất chậm (phải tính toán khoảng cách với toàn bộ dữ liệu trong tập train).

**Model-Based Learning (Học dựa trên mô hình):**
- **Cách hoạt động:** Xây dựng một mô hình toán học tổng quát từ tập dữ liệu huấn luyện và sử dụng mô hình này để dự đoán (VD: Linear Regression, Neural Networks, Decision Trees).
- **Trade-off:** 
  - **Memory:** Rất ít tốn bộ nhớ (chỉ cần lưu các tham số/trọng số của mô hình).
  - **Speed:** Huấn luyện có thể rất lâu (do phải tối ưu hóa hàm mất mát) nhưng dự đoán cực kì nhanh (chỉ cần thay số vào phương trình).

### 3. Chẩn đoán được: Overfitting hay Underfitting? Nguyên nhân? Giải pháp?

**Overfitting (Quá khớp):**
- **Chẩn đoán:** Mô hình hoạt động cực tốt trên tập Train nhưng rất tệ trên tập Validation/Test.
- **Nguyên nhân:** Mô hình quá phức tạp so với lượng dữ liệu, dẫn đến nó "học vẹt" cả những nhiễu (noise) của dữ liệu thay vì học quy luật chung.
- **Giải pháp:**
  - Giảm độ phức tạp mô hình (Regularization: L1, L2, Dropout, giảm số lớp/biến).
  - Thu thập thêm dữ liệu huấn luyện.
  - Xử lý nhiễu/outliers trong dữ liệu tốt hơn.

**Underfitting (Chưa khớp):**
- **Chẩn đoán:** Mô hình hoạt động tệ ở cả tập Train và tập Validation/Test.
- **Nguyên nhân:** Mô hình quá đơn giản để có thể học được các cấu trúc/quy luật tiềm ẩn trong dữ liệu (ví dụ dùng mô hình tuyến tính cho dữ liệu phi tuyến).
- **Giải pháp:**
  - Chọn một mô hình mạnh mẽ và phức tạp hơn.
  - Feature engineering: Tạo thêm các đặc trưng (features) tốt hơn cung cấp cho mô hình.
  - Giảm bớt các ràng buộc của mô hình (VD: giảm hệ số Regularization).

### 4. Kể 3 thách thức về data và 3 thách thức về model

**3 Thách thức về Data (Dữ liệu):**
1. **Thiếu dữ liệu (Insufficient quantity):** Machine learning thường cần rất nhiều dữ liệu để học chính xác (data-hungry).
2. **Dữ liệu kém chất lượng (Poor quality data):** Dữ liệu chứa nhiều lỗi (errors), nhiễu (noise), outliers hoặc bị thiếu giá trị (missing values).
3. **Dữ liệu không mang tính đại diện (Non-representative data):** Nếu tập huấn luyện không phản ánh đúng thực tế, mô hình sẽ không tổng quát hóa được (gây ra sampling bias).

**3 Thách thức về Model (Mô hình):**
1. **Overfitting (Quá khớp):** Mô hình quá phức tạp so với dữ liệu, ghi nhớ nhiễu và mất khả năng tổng quát hóa.
2. **Underfitting (Chưa khớp):** Mô hình quá đơn giản, không học được quy luật ẩn chứa trong dữ liệu.
3. **Irrelevant Features (Đặc trưng không liên quan):** Dữ liệu chứa nhiều đặc trưng không liên quan (rác) làm mô hình khó học hoặc bị đánh lừa (Garbage In, Garbage Out).

### 5. Nếu data bị bias, model của bạn bị ảnh hưởng như thế nào?

- **Dự đoán sai lệch thực tế:** Mô hình học quy luật từ dữ liệu, do đó nếu dữ liệu huấn luyện bị thiên vị (bias), mô hình sẽ tái tạo và thậm chí khuếch đại sự thiên vị đó. Ví dụ: Dữ liệu tuyển dụng trong quá khứ chủ yếu là nam giới, mô hình sẽ tự động đánh giá thấp hồ sơ của nữ giới.
- **Phân biệt đối xử và rủi ro đạo đức:** Các ứng dụng như phê duyệt khoản vay, hệ thống nhận diện khuôn mặt có thể đưa ra quyết định gây bất lợi cho nhóm thiểu số nếu họ bị "under-represented" (thiếu tính đại diện) trong tập dữ liệu.
- **Hiệu suất kém trên nhóm dữ liệu mới:** Mô hình sẽ hoạt động rất tệ khi gặp các nhóm đối tượng không được xuất hiện nhiều trong dữ liệu bị thiên vị (mất tính tổng quát hóa).

### 6. Giải thích được: tại sao cần tách train/validation/test set

Việc chia dữ liệu đảm bảo ta có thể đánh giá đúng khả năng thực sự của mô hình (tính tổng quát hóa - generalization) khi gặp dữ liệu chưa từng thấy.
- **Train set (Tập huấn luyện):** Dùng để cho mô hình học các trọng số và quy luật ẩn chứa. (Học bài)
- **Validation set (Tập kiểm định):** Dùng để điều chỉnh các siêu tham số (hyperparameters), lựa chọn mô hình tốt nhất (model selection) và phát hiện sớm hiện tượng overfitting. Tập này mô hình *không được học trực tiếp* nhưng lại *được nhìn thấy* để điều chỉnh. (Thi thử)
- **Test set (Tập kiểm thử):** Đóng vai trò là "thế giới thực". Dùng chỉ một lần duy nhất vào cuối cùng để đánh giá hiệu suất của mô hình xem khi ra thực tế nó sẽ hoạt động ra sao. Tránh việc "mô hình học thuộc lòng luôn cả đề thi thử (validation)". (Thi thật)

### 7. Mô tả K-Fold Cross-Validation và khi nào dùng

**Mô tả K-Fold Cross-Validation:**
- Quá trình: Chia tập dữ liệu huấn luyện thành $K$ phần nhỏ (folds) bằng nhau. Ta sẽ huấn luyện mô hình $K$ lần. Ở mỗi lần lặp, ta lấy 1 fold làm tập Validation để đánh giá, và $K-1$ folds còn lại làm tập Train.
- Kết quả cuối cùng (độ lỗi/accuracy) sẽ là trung bình của $K$ lần đánh giá đó.

**Khi nào dùng:**
- **Dữ liệu quá ít:** Khi ta không có đủ dữ liệu để trích riêng một tập Validation lớn mà không làm mất đi quá nhiều dữ liệu quý giá dành cho Training. Việc chia K-Fold giúp mọi điểm dữ liệu đều có cơ hội được dùng để train và đánh giá.
- **Muốn đánh giá ổn định hơn (robustness):** Giảm thiểu sự thiên lệch (variance) xảy ra do việc chia Train/Validation ngẫu nhiên (nếu xui chia trúng tập Validation quá dễ hoặc quá khó).

### 8. Phân biệt ML và Statistical Modeling — mục tiêu khác nhau

**Machine Learning (Học máy):**
- **Mục tiêu chính:** Tập trung vào **Dự đoán (Prediction)** và **Hiệu suất (Performance)**. ML không cần mô hình phải dễ giải thích, miễn là khả năng dự đoán trên dữ liệu mới (Test set) càng chính xác càng tốt.
- **Đặc điểm:** Thường dùng cho dữ liệu khổng lồ (Big Data), mô hình cực kỳ phức tạp (như Deep Learning - black box), chú trọng vào việc phát triển thuật toán tự học mà ít có các giả định khắt khe về phân phối dữ liệu.

**Statistical Modeling (Mô hình hóa thống kê):**
- **Mục tiêu chính:** Tập trung vào **Suy luận (Inference)** và **Giải thích (Explainability)**. Cố gắng tìm ra mối quan hệ toán học và nhân quả giữa các biến (Ví dụ: Yếu tố X thay đổi 1 đơn vị thì Y thay đổi thế nào?).
- **Đặc điểm:** Yêu cầu các giả định nghiêm ngặt về dữ liệu (độc lập, phân phối chuẩn...). Thường làm việc với các mẫu dữ liệu nhỏ hơn để chứng minh các giả thuyết hoặc ước lượng.
