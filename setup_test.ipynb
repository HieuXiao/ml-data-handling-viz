# Environment Setup Verification
This notebook verifies the installation and functional integrity of core data science libraries.
* **Libraries:** `numpy`, `pandas`, `matplotlib`, `seaborn`
* **Purpose:** Ensure the local environment is ready for development.

Environment Audit & Validation

```python
import importlib.util
import sys

# Configuration : List of core dependencies to vertify
REQUIRED_PKGS = ["numpy", "pandas", "matplotlib", "seaborn", "jupyter"]

def verify_environment():
    print(f"{'Library':<15} | {'Status':<12} | {'Location'}")
    print("-" * 50)

    missing = []
    for pkg in REQUIRED_PKGS:
        # Check if the package is visible to the current Python interpreter
        spec = importlib.util.find_spec(pkg)
        found = spec is not None

        status = "✅ Installed" if found else "❌ Missing"
        location = spec.origin if found and spec.origin else "N/A"

        print(f"{pkg:<15} | {status:<12} | {location}")

        if not found:
            missing.append(pkg)

    return missing

# Execute audit
missing_packages = verify_environment()

if not missing_packages:
    print("\n" + "="*50)
    print("✨ SUCCESS: Environment is fully configured.")
    print("="*50)
else:
    print("\n" + "!"*50)
    print(f"⚠️ WARNING: {len(missing_packages)} package(s) not found.")
    print(f"Target: {', '.join(missing_packages)}")
    print("!"*50)
```

Automated Dependency Recovery

```python
if missing_packages:
    print(f"Installing missing dependencies: {missing_packages}...")

    # Using %pip magic is the official standard for Jupyter
    # to ensure installation into the correct active kernel.
    %pip install {" ".join(missing_packages)}

    print("\n✅ Installation complete. Please RESTART THE KERNEL to apply changes.")
else:
    print("All dependencies are already satisfied. No action required.")
```
